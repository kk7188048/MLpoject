```python
import json

ROUTER_PROMPT = """You are a query router for an AI agent. Your ONLY job is to classify whether a user's question needs real-time web search or can be answered from training knowledge.

Respond ONLY with valid JSON, nothing else. No explanation, no markdown.

Rules:
- "search" for: current events, prices, news, recent releases, "latest/today/now/current" questions
- "direct" for: concepts, history, math, coding, general knowledge that doesn't change

Output format:
{
  "decision": "search" or "direct",
  "reason": "one line explanation",
  "search_queries": ["query1", "query2"]
}

search_queries should be empty list [] if decision is "direct".
Generate 2-3 focused search queries if decision is "search"."""


def route_query(user_query):
    response = client.chat.completions.create(
        model="moonshotai/kimi-k2-instruct-0905",
        messages=[
            {"role": "system", "content": ROUTER_PROMPT},
            {"role": "user", "content": user_query}
        ],
        max_tokens=200,
        temperature=0.1
    )

    raw = response.choices[0].message.content.strip()

    try:
        result = json.loads(raw)
    except json.JSONDecodeError:
        import re
        match = re.search(r'\{.*\}', raw, re.DOTALL)
        result = json.loads(match.group()) if match else {"decision": "search", "reason": "parse failed", "search_queries": [user_query]}

    return result
```

```python
import trafilatura
from ddgs import DDGS
import time

def search_web(queries, max_results=3):
    all_results = []
    seen_urls = set()

    with DDGS() as ddgs:
        for query in queries:
            results = list(ddgs.text(query, max_results=max_results))
            for r in results:
                if r['href'] not in seen_urls:
                    seen_urls.add(r['href'])
                    all_results.append({
                        'title': r['title'],
                        'url': r['href'],
                        'snippet': r['body']
                    })

    return all_results[:5]


def scrape_url(url):
    try:
        downloaded = trafilatura.fetch_url(url)
        text = trafilatura.extract(downloaded)
        return text[:2000] if text else None
    except:
        return None


def search_and_scrape(queries):
    print(f" Searching for: {queries}")
    sources = search_web(queries)
    print(f"   Found {len(sources)} URLs")

    enriched = []
    for i, source in enumerate(sources):
        print(f"   Scraping [{i+1}/{len(sources)}]: {source['url'][:60]}...")
        content = scrape_url(source['url'])
        time.sleep(0.5)

        enriched.append({
            'title': source['title'],
            'url': source['url'],
            'snippet': source['snippet'],
            'content': content
        })

    scraped_count = sum(1 for s in enriched if s['content'])
    print(f"   Successfully scraped: {scraped_count}/{len(enriched)}")
    return enriched
```

```python
test_route = route_query("What are the latest AI releases in 2025?")
print("Router decision:", test_route['decision'])
print("Search queries:", test_route['search_queries'])
print()

sources = search_and_scrape(test_route['search_queries'])

for s in sources:
    print(f" {s['title']}")
    print(f"   URL     : {s['url'][:70]}")
    print(f"   Scraped : {' ' + str(len(s['content'])) + ' chars' if s['content'] else ' failed'}")
    print()
```

```python
def build_context(sources):
    context = ""
    for i, source in enumerate(sources):
        content = source['content'] or source['snippet']
        context += f"[{i+1}] {source['title']}\nURL: {source['url']}\n{content}\n\n"
    return context


SYNTHESIZER_PROMPT = """You are a research synthesizer. Given a user question and web sources, write a clear and accurate answer.

Rules:
- Cite sources inline using [1], [2], [3] etc.
- Only use information present in the sources, never hallucinate
- Be concise but complete
- If sources contradict each other, mention it
- End with a Sources section listing all URLs used"""


def synthesize(user_query, sources):
    context = build_context(sources)

    user_message = f"""Question: {user_query}

Sources:
{context}

Write a comprehensive answer with inline citations."""

    response = client.chat.completions.create(
        model="moonshotai/kimi-k2-instruct-0905",
        messages=[
            {"role": "system", "content": SYNTHESIZER_PROMPT},
            {"role": "user", "content": user_message}
        ],
        max_tokens=1000,
        temperature=0.3
    )

    return response.choices[0].message.content
```

```python
DIRECT_PROMPT = """You are a knowledgeable assistant. Answer the user's question clearly from your training knowledge. Be concise and accurate."""

def direct_answer(user_query):
    response = client.chat.completions.create(
        model="moonshotai/kimi-k2-instruct-0905",
        messages=[
            {"role": "system", "content": DIRECT_PROMPT},
            {"role": "user", "content": user_query}
        ],
        max_tokens=800,
        temperature=0.4
    )
    return response.choices[0].message.content
```

```python
def ask_the_web(user_query):
    print(f"\n{'='*60}")
    print(f"Query: {user_query}")
    print('='*60)

    print("\n Stage 1: Routing...")
    route = route_query(user_query)
    print(f"   Decision : {route['decision']}")
    print(f"   Reason   : {route['reason']}")

    if route['decision'] == 'search':
        print("\n Stage 2: Searching the web...")
        sources = search_and_scrape(route['search_queries'])

        print("\n Stage 3: Synthesizing answer...")
        answer = synthesize(user_query, sources)
    else:
        print("\n Answering from training knowledge...")
        answer = direct_answer(user_query)
        sources = []

    print(f"\n{'='*60}")
    print("ANSWER")
    print('='*60)
    print(answer)

    return answer, sources
```

```python
%%writefile app.py

import streamlit as st
import json
import time
import re
from groq import Groq
from ddgs import DDGS
import trafilatura
import os

GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
client = Groq(api_key=GROQ_API_KEY)

ROUTER_PROMPT = """You are a query router for an AI agent. Your ONLY job is to classify whether a user's question needs real-time web search or can be answered from training knowledge.

Respond ONLY with valid JSON, nothing else. No explanation, no markdown.

Rules:
- "search" for: current events, prices, news, recent releases, "latest/today/now/current" questions
- "direct" for: concepts, history, math, coding, general knowledge that doesn't change

Output format:
{
  "decision": "search" or "direct",
  "reason": "one line explanation",
  "search_queries": ["query1", "query2"]
}

search_queries should be empty list [] if decision is "direct"""

SYNTHESIZER_PROMPT = """You are a research synthesizer. Given a user question and web sources, write a clear and accurate answer.

Rules:
- Cite sources inline using [1], [2], [3] etc.
- Only use information present in the sources, never hallucinate
- Be concise but complete
- If sources contradict each other, mention it
- End with a Sources section listing all URLs used"""

DIRECT_PROMPT = """You are a knowledgeable assistant. Answer the user's question clearly from your training knowledge. Be concise and accurate."""

def route_query(user_query):
    response = client.chat.completions.create(
        model="moonshotai/kimi-k2-instruct-0905",
        messages=[
            {"role": "system", "content": ROUTER_PROMPT},
            {"role": "user", "content": user_query}
        ],
        max_tokens=200,
        temperature=0.1
    )
    raw = response.choices[0].message.content.strip()
    try:
        return json.loads(raw)
    except:
        match = re.search(r'\\{.*\\}', raw, re.DOTALL)
        return json.loads(match.group()) if match else {"decision": "search", "reason": "parse failed", "search_queries": [user_query]}


def search_web(queries, max_results=3):
    all_results = []
    seen_urls = set()
    with DDGS() as ddgs:
        for query in queries:
            results = list(ddgs.text(query, max_results=max_results))
            for r in results:
                if r['href'] not in seen_urls:
                    seen_urls.add(r['href'])
                    all_results.append({
                        'title': r['title'],
                        'url': r['href'],
                        'snippet': r['body']
                    })
    return all_results[:5]


def scrape_url(url):
    try:
        downloaded = trafilatura.fetch_url(url)
        text = trafilatura.extract(downloaded)
        return text[:2000] if text else None
    except:
        return None


def search_and_scrape(queries):
    sources = search_web(queries)
    enriched = []
    for source in sources:
        content = scrape_url(source['url'])
        time.sleep(0.5)
        enriched.append({
            'title': source['title'],
            'url': source['url'],
            'snippet': source['snippet'],
            'content': content
        })
    return enriched


def build_context(sources):
    context = ""
    for i, source in enumerate(sources):
        content = source['content'] or source['snippet']
        context += f"[{i+1}] {source['title']}\\nURL: {source['url']}\\n{content}\\n\\n"
    return context


def synthesize(user_query, sources):
    context = build_context(sources)
    user_message = f"""Question: {user_query}\\n\\nSources:\\n{context}\\n\\nWrite a comprehensive answer with inline citations."""
    response = client.chat.completions.create(
        model="moonshotai/kimi-k2-instruct-0905",
        messages=[
            {"role": "system", "content": SYNTHESIZER_PROMPT},
            {"role": "user", "content": user_message}
        ],
        max_tokens=1000,
        temperature=0.3
    )
    return response.choices[0].message.content


def direct_answer(user_query):
    response = client.chat.completions.create(
        model="moonshotai/kimi-k2-instruct-0905",
        messages=[
            {"role": "system", "content": DIRECT_PROMPT},
            {"role": "user", "content": user_query}
        ],
        max_tokens=800,
        temperature=0.4
    )
    return response.choices[0].message.content

st.set_page_config(
    page_title="Ask-the-Web Agent",
    page_icon="",
    layout="wide"
)

st.title(" Ask-the-Web Agent")
st.caption("Perplexity-style AI agent • Routes intelligently • Cites sources")

with st.sidebar:
    st.header(" Agent Trace")
    st.caption("Watch the agent think in real time")
    trace_container = st.empty()

st.markdown("**Try these:**")
cols = st.columns(4)
examples = [
    "Latest AI releases in 2025",
    "What is QLoRA?",
    "Current Bitcoin price",
    "How does RAG work?"
]
for i, ex in enumerate(examples):
    if cols[i].button(ex, use_container_width=True):
        st.session_state['query'] = ex

query = st.text_input(
    "Ask anything",
    value=st.session_state.get('query', ''),
    placeholder="e.g. What are the latest AI models in 2025?",
    key="input"
)

if st.button("Ask", type="primary", use_container_width=True) and query:
    trace_log = []

    with st.status(" Routing query...", expanded=True) as status:
        route = route_query(query)
        decision = route['decision']
        icon = "" if decision == "search" else ""
        st.write(f"**Decision:** {icon} {decision.upper()}")
        st.write(f"**Reason:** {route['reason']}")
        trace_log.append(f"**Router** → `{decision}` — {route['reason']}")

        if decision == "search":
            st.write(f"**Search queries:** {route['search_queries']}")

        sources = []
        if decision == "search":
            status.update(label=" Searching the web...")
            sources = search_and_scrape(route['search_queries'])
            scraped = sum(1 for s in sources if s['content'])
            st.write(f"**Sources found:** {len(sources)} | **Scraped:** {scraped}")
            trace_log.append(f"**Searcher** → {len(sources)} sources, {scraped} scraped")

            for s in sources:
                st.write(f"  • [{s['title'][:60]}]({s['url']})")

        if decision == "search":
            status.update(label=" Synthesizing answer...")
            trace_log.append("Synthesizer → generating cited answer")
            answer = synthesize(query, sources)
        else:
            status.update(label=" Answering from knowledge...")
            trace_log.append("Direct LLM → answering from training knowledge")
            answer = direct_answer(query)

        status.update(label="Done!", state="complete")


    trace_container.markdown("\n\n".join(trace_log))

    st.divider()
    st.markdown("### Answer")
    st.markdown(answer)

    if sources:
        st.divider()
        st.markdown("### Sources")
        for i, s in enumerate(sources):
            st.markdown(f"**[{i+1}]** [{s['title']}]({s['url']})")
```
